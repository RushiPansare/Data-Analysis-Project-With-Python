```python
data['Payment_Method'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title("Payment Method use by Customer")
```

```python
data['State'].value_counts()
```

```python
data.groupby('State')['Total_Sales_INR'].sum()
```

```python
plt.figure(figsize=(15,5))
data.groupby('State')['Total_Sales_INR'].sum().plot(kind='bar',color='orange')
plt.xticks(rotation=45)
plt.title("Region wise Sales",pad=20)
```

```python
data['Delivery_Status'].value_counts().plot(kind='bar')
plt.title("Deliver_Status",pad=20)
plt.xticks(rotation=360)
```

```python
data.groupby('Product_Category')['Quantity'].sum()
```

```python
plt.figure(figsize=(10,5))
data.groupby('Product_Category')['Quantity'].sum().plot(kind='bar',color='skyblue')
plt.title("Category wise Sales(In Quantity)",pad=20)
plt.xticks(rotation=360)
```
