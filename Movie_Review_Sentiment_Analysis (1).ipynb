##  Movie Review Sentiment Analysis Projcet 

```python
import pandas as pd
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.corpus import stopwords 
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from wordcloud import WordCloud,STOPWORDS
```

```python
data=pd.read_csv("movie_review.csv")
data
```

```python
data.info()
```

```python
# converting movie review columns into one list   

text_info=[]
for words in data['text']:
    text_info.append(words)

text_info
```

```python
## performing word_tokenize for further sentiment analyssis

list_tok=[]       # here we take again one list beacuse our "text_info " is one list so when we do word tokenize wehave to acess element throug for loop
for text in text_info:
    word_tok=word_tokenize(text)  
    list_tok.append(word_tok)
print(list_tok)       # now this list_tok is one list 

# now we performed the word_tokenization
```

```python
## removed stop words
stop_word=set(stopwords.words("english"))
filtr_word=[]
for sentence in list_tok:
    for words in sentence:          # here we used nested loop beause  in our "list_tok " is one kind of nested list so we required nested loop
        if words not in stop_word:    # to reach in the inside words so we can do comparison so that is why we used two for loop  
            filtr_word.append(words)
        
print("reviews after removed stopwords:", filtr_word)
```

```python
## Sentiment analysis 
sentiment_ana=SentimentIntensityAnalyzer()
for word in filtr_word:
    print(sentiment_ana.polarity_scores(word))
```

```python
import matplotlib.pyplot as plt
text="".join(filtr_word)      # this is join method we have used here to convert our list "filtr_word " into string format .beause if want to diplay a wordcloud then we have to pass string to wordcloud. we could not able to pass lst,set,tuples.
wc=WordCloud(collocations=False,background_color='black',width=800,height=400).generate(text)
plt.imshow(wc,interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud of Movie Review")
plt.show()
```
