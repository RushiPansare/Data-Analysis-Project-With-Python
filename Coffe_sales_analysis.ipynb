##  Coffe Sales Data Analysis Project

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv("Coffe_Sales.csv")
data
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
data['coffee_name'].value_counts()
```

```python
data.groupby('coffee_name')['money'].sum()
```

```python
data.groupby('coffee_name')['money'].sum().plot(kind='bar',color='skyblue')
plt.xticks(rotation=45)
plt.title("Total Revenue of per copy",pad=20)
plt.show()
```

#### Most Selling coffie type is Latte & Second most Selling cofee is " Americano with milk and  3rd one is" Cappuccino. 

```python
data['cash_type'].value_counts()
```

```python
data.groupby('Weekday')['money'].sum().plot(kind='pie',autopct='%1.1f%%')
```

```python
label=['Jan','feb','March','April','May','June','July','Aug','Sep','Oct','Nov','Dec']
plt.figure(figsize=(10,5))
data.groupby('Month_name')['money'].sum().plot(kind='line',marker='o')
plt.title("Month Wise Sales",pad=20)

plt.grid(ls='--')
```

```python
plt.figure(figsize=(20,5))
data.groupby('Date')['money'].sum().plot(kind='line',color='orange')
plt.title('Year Wise Sales',pad=20)
```
