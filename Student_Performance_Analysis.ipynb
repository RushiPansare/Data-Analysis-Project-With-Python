```python
# conversion of datatype of features  into integer datatype

data['study_hours']=data['study_hours'].astype('int')
data['overall_score']=data['overall_score'].astype('int')
```

```python
data.info()
```

```python
sns.scatterplot(data=data,x='study_hours',y='overall_score')
plt.title("Study_hr vs Overall_score",pad=20)
```

#### positive Relationship(correlation):
      As study hours increases overall_score also increases so above graphs showing high positive correlation between 
       study_hr & Overall_score

```python
plt.figure(figsize=(15,5))
sns.scatterplot(data=data,x='attendance_percentage',y='overall_score',color='orange')
plt.title('Attedance_per vs Overall_Score',pad=20)
```

### Positive Relationship:
  As we can see  above graph also show slightly positive relationship between attendance & overall_score.
 Beacuse as attendance increases overall_score also increses
