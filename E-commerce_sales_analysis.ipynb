```python
print("Most Selling Category ")

data['Category'].value_counts()
```

```python
sns.countplot(data=data,x='Category',)
plt.title("Category wise Demand",pad=20)
plt.grid(ls='--',color='gray')
```

## Interpretation
  As we can see in above graph we have find out which category has highest demands in our.we found which catgeory was slod out mostly .

```python
data['Product Name'].value_counts()
```

```python
data['Product Name'].value_counts().plot(kind='bar',color='orange')
plt.xticks(rotation=40)
```

```python
## Category wise sales

data.groupby('Category')[('Sales')].sum()
```

```python
data.groupby('Category')['Sales'].sum().plot(kind='bar',color=['orange','blue','yellow'],)
plt.title("Category Wise Sales",pad=20)
plt.xticks(rotation=360)
```

### Category wise sales
     
     1) Electronic (This category done Highest sales in dataset)
     2) Accessories (Second most selling category")
     3) Office (least selling category
     

```python
## Region wise sales
data.groupby('Region')[('Sales')].sum()
```

```python
## Region wise sales
data.groupby('Region')[('Sales')].sum().plot(kind='bar',color='maroon')
plt.title('Region Wise sales',pad=20)
plt.xticks(rotation=360)
```

## Region wise sales

    1) West Region has highest Sales 
    2) East - Second most selling region
    3) South-Third most selling region
       
 


```python
# Category Wise Analysis
data.groupby('Category')[['Profit','Sales','Quantity']].sum()
```

```python
data.groupby('Category')['Profit'].sum().plot(kind='pie',autopct='%1.1f%%',labels=['Accessories','Electonic','Office'])
plt.title("Contibution Of Each Category in Profit",pad=20)
plt.legend(loc='lower right')
```

### Conclusion : 
    1) Electronic Category Give "50%" Profit
    2) Accessories Give 39.9%"
    3) office give '10%'
